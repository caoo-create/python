一、字符串的乐趣 (String Fun)
1. 回文判断 is_palindrome
思路：忽略空格和大小写，可以直接将字符串转换为小写并移除所有空格，然后比较正向和反向的字符串。

```python
def is_palindrome(s):
    # 忽略大小写并移除空格
    s_clean = s.lower().replace(" ", "")
    # 比较正序和倒序是否相同
    return s_clean == s_clean[::-1]

# 测试
print(is_palindrome("taco cat"))  # True
print(is_palindrome("T A C O cat")) # True
print(is_palindrome("hello"))     # False
```

2. 字母顺序判断 is_abecedarian
思路：同样忽略空格，将字符串转为小写后，判断相邻字符是否满足前一个字符小于等于后一个字符。

```python
def is_abecedarian(s):
    s_clean = s.lower().replace(" ", "")
    # 遍历并比较相邻字符
    for i in range(len(s_clean) - 1):
        if s_clean[i] > s_clean[i+1]:
            return False
    return True

# 测试
print(is_abecedarian("adder"))      # True
print(is_abecedarian("abcd efgh xyz")) # True
print(is_abecedarian("dog"))        # False
```

3. 移除元音 remove_vowels
思路：使用列表推导式，只保留不在元音集合中的字符。

```python
def remove_vowels(s):
    vowels = "aeiouAEIOU"
    # 过滤掉元音字符，然后拼接成字符串
    return "".join([char for char in s if char not in vowels])

# 测试
print(remove_vowels('cat'))   # 'ct'
print(remove_vowels('audio')) # 'd'
```

二、列表的乐趣 (List Fun)
1. 列表反转 list_reverse
思路：首先检查输入是否为列表，如果不是则抛出 TypeError，如果是则返回反转后的新列表。

```python
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表 (list)")
    return lst[::-1]

# 测试
print(list_reverse([1, 2, 3])) # [3, 2, 1]
try:
    list_reverse("not a list")
except TypeError as e:
    print("Error:", e)
```

2. 序列排序检查 is_sorted
要求：只能遍历一次序列。
思路：使用 Python 内置的 all() 函数结合生成器表达式，或者在单次循环中检查相邻元素。

```python
def is_sorted(seq):
    # 使用 all() 配合生成器，在底层只需遍历一次
    # 检查前一个元素是否始终小于等于后一个元素
    return all(seq[i] <= seq[i+1] for i in range(len(seq)-1))

# 测试
print(is_sorted([1, 2, 3, 5, 5])) # True
print(is_sorted([1, 3, 2, 4]))    # False
```

三、作为向量的元组 (Tuples as Vectors)
1. 向量标量乘法 vec_scalar_mult
思路：检查输入元组和标量的类型，返回一个新的元组。

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple) or not isinstance(s, (int, float)):
        raise TypeError("第一个参数必须是元组，第二个参数必须是整数或浮点数")
    if not all(isinstance(x, (int, float)) for x in t):
        raise TypeError("元组内的元素必须是整数或浮点数")
    return tuple(x * s for x in t)

# 测试
print(vec_scalar_mult((1, 2, 3), 2.5)) # (2.5, 5.0, 7.5)
```

3. 矩阵合法性检查 check_valid_mx
思路：检查外层元组，再遍历每一行检查是否为元组、长度是否一致且元素为数字。

```python
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if not all(isinstance(row, tuple) for row in mx):
        return False
    if len(mx) == 0:
        return True # 视空元组为空矩阵，视为合法
    
    row_len = len(mx[0])
    for row in mx:
        if len(row) != row_len:
            return False
        if not all(isinstance(x, (int, float)) for x in row):
            return False
    return True

# 测试
print(check_valid_mx(((1, 2), (3, 4)))) # True
print(check_valid_mx(((1, 2), (3, 4, 5)))) # False
```

4. 矩阵与向量相乘 mx_vec_mult
思路：先用 check_valid_mx 检查矩阵，再检查向量类型和维度，最后计算结果。

```python
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise ValueError("输入的不是合法的矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    if len(mx[0]) != len(vec):
        raise ValueError("矩阵的列数必须等于向量的长度")
    
    result = []
    for row in mx:
        # 每行与向量求内积
        result.append(sum(x * y for x, y in zip(row, vec)))
    return tuple(result)

# 测试
print(mx_vec_mult(((1, 2), (3, 4)), (5, 6))) # (17, 39)
```

四、更多向量的乐趣 (OOP - 面向对象)
1 & 2. 定义 Vector 类与属性选择
解释：维度和元素应该是实例属性（Instance Attributes）。因为每个具体的向量对象都有自己独立的维度和坐标值，如果作为类属性（Class Attributes），所有向量对象将共享同一份数据，这不符合逻辑。

```python
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or dim < 0:
            raise ValueError("维度必须是非负整数")
        self.dim = dim
        if elements is None:
            self.elements = tuple([0] * dim)
        else:
            if not isinstance(elements, (list, tuple)):
                raise TypeError("元素必须是列表或元组")
            if len(elements) != dim:
                raise ValueError("元素个数必须与维度一致")
            if not all(isinstance(x, (int, float)) for x in elements):
                raise TypeError("元素必须是整数或浮点数")
            self.elements = tuple(elements) 

    # --- 这里是补全的比较方法 ---
    
    def _check_dim(self, other):
        # 辅助函数：检查比较对象是否合法
        if not isinstance(other, Vector):
            raise TypeError("比较对象必须是 Vector")
        if self.dim != other.dim:
            raise ValueError("无法比较维度不同的向量")

    def __eq__(self, other): # 等于 ==
        self._check_dim(other)
        return self.elements == other.elements

    def __lt__(self, other): # 小于 <
        self._check_dim(other)
        return self.elements < other.elements

    def __le__(self, other): # 小于等于 <=
        self._check_dim(other)
        return self.elements <= other.elements

    def __gt__(self, other): # 大于 >
        self._check_dim(other)
        return self.elements > other.elements

    def __ge__(self, other): # 大于等于 >=
        self._check_dim(other)
        return self.elements >= other.elements

    # --- 下面是其他方法 ---
    
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是 Vector 对象")
        if self.dim != other.dim:
            raise ValueError("维度不一致，无法计算内积")
        return float(sum(x * y for x, y in zip(self.elements, other.elements)))

    def __mul__(self, other):
        if isinstance(other, (int, float)):
            return Vector(self.dim, tuple(x * other for x in self.elements))
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("维度不一致，无法进行逐元素相乘")
            return Vector(self.dim, tuple(x * y for x, y in zip(self.elements, other.elements)))
        else:
            raise TypeError("不支持的类型相乘")

    def __rmul__(self, other):
        return self.__mul__(other)
    
    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"
```

5. 乘法运算符重载
思路：处理标量乘法（左乘/右乘）和向量逐元素乘法

```python
def __mul__(self, other):
        # 标量乘法：Vector * 数字
        if isinstance(other, (int, float)):
            return Vector(self.dim, tuple(x * other for x in self.elements))
        # 向量乘法：Vector * Vector
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("维度不一致，无法进行逐元素相乘")
            return Vector(self.dim, tuple(x * y for x, y in zip(self.elements, other.elements)))
        else:
            raise TypeError("不支持的类型相乘")

def __rmul__(self, other):
        # 处理 数字 * Vector (右乘)
        return self.__mul__(other)
    
    # 方便打印查看
def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"
```
