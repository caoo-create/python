## 实验 1：正则表达式的基础匹配

```python
import re

print("="*20, "实验 1", "="*20)

# 任务 1
text1 = "cat on mat; later: cats and dogs"
print("--- 任务 1 ---")
print("匹配数字 \\d+ :", re.findall(r"\d+", text1))
print("匹配空白 \\s+ :", re.findall(r"\s+", text1))
print("匹配独立单词 \\bcat\\b :", re.findall(r"\bcat\b", text1)) 
# 注：cats不会被匹配，因为\b是单词边界

# 任务 2
text2 = "cat dog bird CAT"
print("\n--- 任务 2 ---")
print("匹配 cat|dog :", re.findall(r"cat|dog", text2))
# 区分大小写，所以 CAT 不会被匹配到

# 演示匹配差异
print("re.findall(r'a|aa|aaa', 'aaaa') 结果:", re.findall(r"a|aa|aaa", "aaaa"))
print("re.findall(r'aaa|aa|a', 'aaaa') 结果:", re.findall(r"aaa|aa|a", "aaaa"))
# 解释：正则表达式引擎是从左到右扫描的。第一行先尝试匹配 'a'，成功后继续匹配下一个 'a'。
# 第二行先尝试匹配最长的 'aaa'，成功后再匹配剩下的 'a'。

# 任务 3
text3 = "ct cat caat caaat"
print("\n--- 任务 3 ---")
print("ca*t    :", re.findall(r"ca*t", text3))
print("ca+t    :", re.findall(r"ca+t", text3))
print("ca?t    :", re.findall(r"ca?t", text3))
print("ca{2}t  :", re.findall(r"ca{2}t", text3))
print("ca{1,2}t:", re.findall(r"ca{1,2}t", text3))
"""
量词含义说明：
* : 匹配前面的字符 0 次或多次 (所以 ct 也被匹配)
+ : 匹配前面的字符 1 次或多次 (排除 ct)
? : 匹配前面的字符 0 次或 1 次 (只匹配 ct 和 cat)
{n} : 精确匹配前面的字符 n 次 (只匹配 caat)
{n,m} : 匹配前面的字符 n 到 m 次 (匹配 cat 和 caat)
"""
```

## 实验 2：分组、回溯引用与替换

```python
import re

print("\n" + "="*20, "实验 2", "="*20)

# 任务 1
text_email = "联系：alice@example.com；备选：bob@test.org"
print("--- 任务 1 ---")
# 使用捕获组 () 提取用户名和域名
emails = re.findall(r"([\w.-]+)@([\w.-]+)", text_email)
print("提取结果 (用户名, 域名):", emails)

# 任务 2
text_repeat = "cat cat, dog cat, bird bird"
print("\n--- 任务 2 ---")
# \b(\w+)\s+\1\b : 捕获组(\w+)匹配一个单词，\s+匹配空格，\1 引用前面的捕获组内容
repeats = re.findall(r"\b(\w+)\s+\1\b", text_repeat)
print("连续重复的单词:", repeats)

# 任务 3
text_date = "2024-05-01\n2024/05/02\n2024.05.03"
print("\n--- 任务 3 ---")
print("转换前文本:\n" + text_date)
# 捕获组：(\d{4}) 年, (\d{2}) 月, (\d{2}) 日。使用 [-/.] 匹配分隔符
# 替换表达式：\2-\3-\1 即 月-日-年
converted_date = re.sub(r"(\d{4})[-/.](\d{2})[-/.](\d{2})", r"\2-\3-\1", text_date)
print("\n转换后文本:\n" + converted_date)

# 任务 4
text_phone = "联系我：123-456-7890 或 987.654.3210"
print("\n--- 任务 4 ---")
# 捕获三个数字段，替换为 (段1) 段2-段3
formatted_phone = re.sub(r"(\d{3})[-.](\d{3})[-.](\d{4})", r"(\1) \2-\3", text_phone)
print("格式化后:", formatted_phone)
```

## 实验 3：使用 BeautifulSoup 解析 HTML

```python
from bs4 import BeautifulSoup

print("\n" + "="*20, "实验 3", "="*20)

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

# 构建解析树
soup = BeautifulSoup(html_doc, "html.parser")

# 任务 1
print("--- 任务 1 ---")
print("页面标题文本:", soup.title.get_text())
h1_tag = soup.h1
print(f"h1 的标签名: {h1_tag.name}, 文本: {h1_tag.get_text()}")

# 任务 2
print("\n--- 任务 2 ---")
# 找出所有 class="course" 的段落
courses = soup.find_all('p', class_='course')
print("所有 course 段落:")
for c in courses:
    print(" -", c.get_text())

# 找出所有链接，并输出文字和 href
links = soup.find_all('a')
print("所有链接:")
for a in links:
    print(f" - 文本: {a.get_text()}, 属性 href: {a.get('href')}")

# 任务 3
print("\n--- 任务 3 ---")
print("整个文档可读文本:\n", soup.get_text(strip=True)) # strip=True 去除多余换行和空格
print("\n单个段落文本:", soup.p.get_text())
print("单个链接文本:", soup.a.get_text())
# 对比说明：
# 整个文档会提取所有标签内的所有文本（包括 html, head, body 等，但 get_text 默认会剥离标签）。
# 单个段落/链接则只提取特定标签包裹的文本。
```
