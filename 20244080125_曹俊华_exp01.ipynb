编写score_level(score)函数：90～100分返回“优秀”，80～不足90分返回
   “良好”，60～不足80分返回“及格”，0～不足60分返回“不及格”。对于超出0～
   100范围的数值，抛出ValueError。

```python
def score_level(score):
   #范围检查：超出 0~100 抛 ValueError
    if score < 0 or score > 100:
        raise ValueError(f"分数必须在 0~100 之间，收到 {score}")

    if score >= 90:   return "优秀"
    elif score >= 80: return "良好"
    elif score >= 60: return "及格"
    else:             return "不及格"

test_cases = [-1, -0.1, 0, 59.9, 60, 99.9, 100, 100.1, 101]

for s in test_cases:
    try:
        result = score_level(s)
        print(f"{s:>6} -> {result}")
    except ValueError as e:
        print(f"{s:>6} -> ValueError: {e}")
```

## 2.3　
编写递归版和迭代版阶乘函数，约定0! = 1，输入为非负整数。比较n = 0、1、
5、10时的结果。

```python
import math


def factorial_recursive(n):
    """递归版阶乘。约定 0! = 1,输入必须是非负整数。"""
    if isinstance(n, bool) or not isinstance(n, int):
        raise ValueError(f"n 必须是非负整数，收到 {type(n).__name__}: {n!r}")
    if n < 0:
        raise ValueError(f"n 必须是非负整数，收到 {n}")
    if n == 0 or n == 1:          # 递归终止条件
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    """迭代版阶乘。约定 0! = 1，输入必须是非负整数。"""
    if isinstance(n, bool) or not isinstance(n, int):
        raise ValueError(f"n 必须是非负整数，收到 {type(n).__name__}: {n!r}")
    if n < 0:
        raise ValueError(f"n 必须是非负整数，收到 {n}")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result

# 比较 n = 0, 1, 5, 10 的结果
print(f"{'n':<4}{'递归版':<12}{'迭代版':<12}{'math.factorial':<16}{'三者一致'}")
print("-" * 55)
for n in [0, 1, 5, 10]:
    r = factorial_recursive(n)
    it = factorial_iterative(n)
    m = math.factorial(n)
    print(f"{n:<4}{r:<12}{it:<12}{m:<16}{r == it == m}")# 比较 n = 0, 1, 5, 10 的结果
print(f"{'n':<4}{'递归版':<12}{'迭代版':<12}{'math.factorial':<16}{'三者一致'}")
print("-" * 55)
for n in [0, 1, 5, 10]:
    r = factorial_recursive(n)
    it = factorial_iterative(n)
    m = math.factorial(n)
    print(f"{n:<4}{r:<12}{it:<12}{m:<16}{r == it == m}")

# 异常测试
for bad in [-1, 2.5, "5"]:
    try:
        factorial_recursive(bad)
    except ValueError as e:
        print(f"递归版 {bad!r:>5} -> ValueError: {e}")
```

## 2.4　
编写summary(values)函数，接收非空数值列表，以元组返回最小值、最大值和平
  均值，并通过元组解包接收结果。

```python
def summary(values):
    """接收非空数值列表，返回 (最小值, 最大值, 平均值) 三元组。"""
    if not values:
        raise ValueError("values 不能为空列表")
    return min(values), max(values), sum(values) / len(values)

data = [85, 58, 92, 76, 60, 89]

# 元组解包接收
lowest, highest, average = summary(data)
print(f"最小值 = {lowest}")
print(f"最大值 = {highest}")
print(f"平均值 = {average:.2f}")

# 验证返回值确实是元组
result = summary(data)
print("\n返回值类型:", type(result).__name__)
print("返回值内容:", result)
print("可以下标访问 result[0] =", result[0])

# 空列表会抛 ValueError
try:
    summary([])
except ValueError as e:
    print("捕获到 ValueError:", e)
```

编写sum_all(*args)函数，返回所有参数之和；分别测试无参数、单个参数和多
个参数的情况。

```python
def sum_all(*args):
    """返回所有位置参数之和；无参数时返回 0。"""
    return sum(args)


def show_args(*args):
    """演示 *args 打包后的类型。"""
    print(f"  收到 {len(args)} 个参数, 类型={type(args).__name__}, args={args}")

# 无参数
print("无参数:")
show_args()
print("  sum_all() =", sum_all())

# 单个参数
print("\n单个参数:")
show_args(7)
print("  sum_all(7) =", sum_all(7))

# 多个参数
print("\n多个参数:")
show_args(1, 2, 3, 4, 5)
print("  sum_all(1, 2, 3, 4, 5) =", sum_all(1, 2, 3, 4, 5))

# 把列表解包后传入
nums = [10, 20, 30]
print("\n列表解包传入:")
show_args(*nums)
print("  sum_all(*[10, 20, 30]) =", sum_all(*nums))
```

```python
import os
import sys


def locate_notebook_dir(notebook_name):
    """定位 notebook 所在目录，返回绝对路径；找不到则返回 None。"""
    here = os.path.abspath(os.getcwd())

    # 情况 1：当前目录就是 notebook 所在目录
    if os.path.exists(os.path.join(here, notebook_name)):
        return here

    # 情况 2：当前目录是 notebook 所在目录的父目录
    try:
        for entry in os.listdir(here):
            candidate = os.path.join(here, entry)
            if os.path.isdir(candidate) and os.path.exists(os.path.join(candidate, notebook_name)):
                return candidate
    except OSError:
        pass

    # 情况 3：向上逐级查找
    current = here
    while True:
        if os.path.exists(os.path.join(current, notebook_name)):
            return current
        parent = os.path.dirname(current)
        if parent == current:
            return None
        current = parent


NOTES = "20244080125_曹俊华_exp01.ipynb"
target_dir = locate_notebook_dir(NOTES)

if target_dir is None:
    print("警告：未能定位 notebook 所在目录，将使用当前目录：", os.getcwd())
elif target_dir != os.path.abspath(os.getcwd()):
    os.chdir(target_dir)
    print("已切换到 notebook 所在目录:", target_dir)
else:
    print("工作目录正确:", target_dir)
```

## 4.3　
编写parse_student(line)函数：将成绩转换为整数，检查字段数量、姓名是否
为空及成绩是否在0～100范围内。无效记录抛出ValueError。

```python
def parse_student(line):
    """把一行 "name,score" 解析成 {"name": ..., "score": ...}。

    无效记录一律抛出 ValueError，并说明无效原因。
    """
    parts = line.strip().split(",")

    # 校验 1：字段数量
    if len(parts) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(parts)}")

    name = parts[0].strip()
    raw_score = parts[1].strip()

    # 校验 2：姓名非空
    if not name:
        raise ValueError("姓名不能为空")

    # 校验 3：成绩必须能转成整数
    try:
        score = int(raw_score)
    except ValueError:
        raise ValueError(f"成绩不是合法整数: {raw_score!r}")

    # 校验 4：成绩范围
    if not (0 <= score <= 100):
        raise ValueError(f"成绩超出 0~100 范围: {score}")

    return {"name": name, "score": score}

# 单独测试这个函数
test_lines = ["Alice,85", "Bob,58", "Grace,abc", "Henry,105", "Incomplete", ",90", "  Eva , 60 "]
for line in test_lines:
    try:
        print(f"{line!r:<16} -> {parse_student(line)}")
    except ValueError as e:
        print(f"{line!r:<16} -> ValueError: {e}")
```

```python
valid_records = []   # 有效记录：字典组成的列表
errors = []          # 无效记录：(行号, 原始内容, 错误原因)

with open(DATA_FILE, "r", encoding="utf-8") as f:
    next(f)                                       # 跳过表头
    for line_no, line in enumerate(f, start=2):   # 数据从第 2 行开始
        if not line.strip():                      # 跳过空行
            continue
        try:
            valid_records.append(parse_student(line))
        except ValueError as e:
            errors.append((line_no, line.rstrip("\n"), str(e)))

print(f"有效记录数: {len(valid_records)}")
for r in valid_records:
    print("  ", r)

print(f"\n无效记录数: {len(errors)}")
for line_no, raw, reason in errors:
    print(f"  第{line_no}行 {raw!r:<16} -> {reason}")
```

# 实验 5　类、继承与对象复制

## 5.1　
定义Student类，在__init__()中初始化name和score属性，并实现is_passed()
   和__str__()方法。

```python
class Student:
    """学生对象：姓名 + 成绩。"""

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        """成绩 >= 60 视为及格。"""
        return self.score >= 60

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return f"Student(name={self.name!r}, score={self.score}, {status})"

    def __repr__(self):
        return f"Student({self.name!r}, {self.score})"
    # 简单测试
s1 = Student("Alice", 85)
s2 = Student("Bob", 58)
print(s1)
print(s2)
print("s1.is_passed() =", s1.is_passed(), " s2.is_passed() =", s2.is_passed())
print("直接查看对象:", s1.__dict__)
```

## 5.3　
定义继承自Student的GraduateStudent类，增加research_topic属性，通过
   super().__init__()初始化父类属性，并重写__str__()。

```python
class GraduateStudent(Student):
    """研究生：在学生基础上增加研究方向。"""

    def __init__(self, name, score, research_topic):
        super().__init__(name, score)      # 初始化父类属性
        self.research_topic = research_topic

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return (f"GraduateStudent(name={self.name!r}, score={self.score}, "
                f"research_topic={self.research_topic!r}, {status})")
gs = GraduateStudent("Alice", 85, "机器学习")
print(gs)
print("\nis_passed() 继承自父类:", gs.is_passed())
print("是 GraduateStudent 实例?", isinstance(gs, GraduateStudent))
print("是 Student 实例?       ", isinstance(gs, Student))
print("属性字典:", gs.__dict__)
print("方法解析顺序 (MRO):", [c.__name__ for c in GraduateStudent.__mro__])
# 父类与子类对象放在一起，多态调用 __str__
people = [
    Student("Bob", 58),
    GraduateStudent("Alice", 85, "机器学习"),
    GraduateStudent("Carol", 92, "数据挖掘"),
]

for p in people:
    print(f"{type(p).__name__:<16} {p}")
```

## 5.4　
构造包含嵌套列表的对象，例如{"name": "Alice", "scores": [85, 90]}。分
别进行直接赋值、copy.copy()和copy.deepcopy()，修改内部列表，比较结果。
每次比较都从新建的原始对象开始。构造包含嵌套列表的对象，例如{"name": "Alice", "scores": [85, 90]}。分
别进行直接赋值、copy.copy()和copy.deepcopy()，修改内部列表，比较结果。
每次比较都从新建的原始对象开始。

```python
import copy
def make_original():
    """每次调用都返回一个全新的原始对象（含嵌套列表）。"""
    return {"name": "Alice", "scores": [85, 90]}
print("原始对象工厂产物:", make_original())
# ---- 1) 直接赋值 ----
original = make_original()
alias = original                     # 只是同一个对象的另一个名字
alias["scores"].append(100)          # 修改内部列表
print("【直接赋值 alias = original】")
print("  original:", original)
print("  alias   :", alias)
print("  alias is original          ->", alias is original)
print("  alias['scores'] is original['scores'] ->", alias["scores"] is original["scores"])
# ---- 2) 浅拷贝 copy.copy() ----
original = make_original()           # 重新新建，避免受上一次影响
shallow = copy.copy(original)
shallow["scores"].append(100)        # 修改内部列表
print("【浅拷贝 shallow = copy.copy(original)】")
print("  original:", original)
print("  shallow :", shallow)
print("  shallow is original          ->", shallow is original)
print("  shallow['scores'] is original['scores'] ->", shallow["scores"] is original["scores"])
# ---- 3) 深拷贝 copy.deepcopy() ----
original = make_original()           # 再次重新新建
deep = copy.deepcopy(original)
deep["scores"].append(100)           # 修改内部列表
print("【深拷贝 deep = copy.deepcopy(original)】")
print("  original:", original)
print("  deep    :", deep)
print("  deep is original          ->", deep is original)
print("  deep['scores'] is original['scores'] ->", deep["scores"] is original["scores"])
```

```python
# 汇总对比表
print(f"{'方式':<14}{'外层同一对象':<14}{'内层同一对象':<14}{'改内层是否影响原对象'}")
print("-" * 62)

for label, maker in [
    ("直接赋值", lambda o: o),
    ("copy.copy", copy.copy),
    ("copy.deepcopy", copy.deepcopy),
]:
    original = make_original()
    other = maker(original)
    outer_same = other is original
    inner_same = other["scores"] is original["scores"]
    other["scores"].append(100)
    affected = original["scores"] != [85, 90]
    print(f"{label:<14}{str(outer_same):<14}{str(inner_same):<14}{affected}")
```
