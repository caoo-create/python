## 2.1　`score_level(score)` 成绩等级

边界约定：「90〜100 优秀」表示**闭区间**，所以 `score >= 90`；
其余依次为 `>= 80`、`>= 60`、`< 60`。

超出 0~100 抛 `ValueError`。

````python
def score_level(score):
   #范围检查：超出 0~100 抛 ValueError
    if score < 0 or score > 100:
        raise ValueError(f"分数必须在 0~100 之间，收到 {score}")

    if score >= 90:   return "优秀"
    elif score >= 80: return "良好"
    elif score >= 60: return "及格"
    else:             return "不及格"


# 范围检查测试：含越界、边界点、正常分
test_cases = [-1, -0.1, 0, 59.9, 60, 99.9, 100, 100.1, 101]

for s in test_cases:
    try:
        result = score_level(s)
        print(f"{s:>6} -> {result}")
    except ValueError as e:
        print(f"{s:>6} -> ValueError: {e}")
````

## 2.2　`for` 与 `while` 计算 1~100 的和

- `for` 适合**已知次数 / 可迭代对象**的遍历，用 `range(1, 101)` 生成 1~100
- `while` 靠**条件**控制，必须自己维护计数器，注意别写成死循环

`range(1, 101)` 的右边界取不到，所以要写 101 才能包含 100。

````python
# for 版本
total_for = 0
for i in range(1, 101):
    total_for += i

# while 版本
total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print("for   计算结果:", total_for)
print("while 计算结果:", total_while)
print("内置 sum 验证    :", sum(range(1, 101)))
````

## 2.3　递归版与迭代版阶乘（0! = 1）

- 递归版：`n! = n * (n-1)!`，**必须有终止条件** `n == 0 or n == 1`
- 迭代版：从 2 循环乘到 n，`result` 初值取 1（这样 n=0、n=1 都天然得 1）

约定输入为非负整数，其它情况抛 `ValueError`。

````python
import math


def factorial_recursive(n):
    """递归版阶乘。约定 0! = 1,输入必须是非负整数。"""
    if isinstance(n, bool) or not isinstance(n, int):
        raise ValueError(f"n 必须是非负整数，收到 {type(n).__name__}: {n!r}")
    if n < 0:
        raise ValueError(f"n 必须是非负整数，收到 {n}")
    if n == 0 or n == 1:          # 递归终止条件
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    """迭代版阶乘。约定 0! = 1，输入必须是非负整数。"""
    if isinstance(n, bool) or not isinstance(n, int):
        raise ValueError(f"n 必须是非负整数，收到 {type(n).__name__}: {n!r}")
    if n < 0:
        raise ValueError(f"n 必须是非负整数，收到 {n}")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result
````

````python
# 比较 n = 0, 1, 5, 10 的结果
print(f"{'n':<4}{'递归版':<12}{'迭代版':<12}{'math.factorial':<16}{'三者一致'}")
print("-" * 55)
for n in [0, 1, 5, 10]:
    r = factorial_recursive(n)
    it = factorial_iterative(n)
    m = math.factorial(n)
    print(f"{n:<4}{r:<12}{it:<12}{m:<16}{r == it == m}")
````

````python
# 异常测试
for bad in [-1, 2.5, "5"]:
    try:
        factorial_recursive(bad)
    except ValueError as e:
        print(f"递归版 {bad!r:>5} -> ValueError: {e}")

# 补充：递归深度限制（Python 默认约 1000 层）
import sys
print("\n递归深度上限:", sys.getrecursionlimit())
try:
    factorial_recursive(2000)
except RecursionError as e:
    print("递归过深 -> RecursionError:", e)
    print("所以大 n 应该用迭代版。")

# 迭代版可以算很大的 n（这里用 500 演示）
big = factorial_iterative(500)
print("\n迭代版 500! 的位数:", len(str(big)))
print("Python 对 int->str 的位数限制:", sys.get_int_max_str_digits())
````

## 2.4　`summary(values)`：以元组返回最小值、最大值、平均值

返回多个值时，Python 实际返回的是一个**元组**，可以直接用元组解包接收：

```python
low, high, avg = summary(data)
```

要求列表非空，否则 `min()` / `max()` 本身就会报错，这里主动抛 `ValueError` 让错误更清楚。

````python
def summary(values):
    """接收非空数值列表，返回 (最小值, 最大值, 平均值) 三元组。"""
    if not values:
        raise ValueError("values 不能为空列表")
    return min(values), max(values), sum(values) / len(values)
````

````python
data = [85, 58, 92, 76, 60, 89]

# 元组解包接收
lowest, highest, average = summary(data)
print(f"最小值 = {lowest}")
print(f"最大值 = {highest}")
print(f"平均值 = {average:.2f}")

# 验证返回值确实是元组
result = summary(data)
print("\n返回值类型:", type(result).__name__)
print("返回值内容:", result)
print("可以下标访问 result[0] =", result[0])
````

````python
# 空列表会抛 ValueError
try:
    summary([])
except ValueError as e:
    print("捕获到 ValueError:", e)
````

## 2.5　`sum_all(*args)`：可变位置参数

`*args` 把任意个位置参数**打包成一个元组**。所以：

- 无参数 → `args == ()` → 和为 0
- 单参数 → `args == (7,)`
- 多参数 → 正常累加

`sum(())` 的结果是 0，所以直接 `return sum(args)` 就能覆盖无参数的情况。

````python
def sum_all(*args):
    """返回所有位置参数之和；无参数时返回 0。"""
    return sum(args)


def show_args(*args):
    """演示 *args 打包后的类型。"""
    print(f"  收到 {len(args)} 个参数, 类型={type(args).__name__}, args={args}")
````

````python
# 无参数
print("无参数:")
show_args()
print("  sum_all() =", sum_all())

# 单个参数
print("\n单个参数:")
show_args(7)
print("  sum_all(7) =", sum_all(7))

# 多个参数
print("\n多个参数:")
show_args(1, 2, 3, 4, 5)
print("  sum_all(1, 2, 3, 4, 5) =", sum_all(1, 2, 3, 4, 5))

# 把列表解包后传入
nums = [10, 20, 30]
print("\n列表解包传入:")
show_args(*nums)
print("  sum_all(*[10, 20, 30]) =", sum_all(*nums))
````

## 3.1　用 `zip()` 构建字典并遍历

`zip()` 把多个可迭代对象**按位置配对**成元组，`dict()` 可以直接把
「键值对的可迭代对象」转成字典。

````python
pairs = list(zip(names, scores))
print("zip() 配对结果:", pairs)

score_dict = dict(zip(names, scores))
print("\n构建的字典:", score_dict)

print("\n使用 items() 逐项输出:")
for name, score in score_dict.items():
    print(f"  {name:<6} {score}")
````

## 3.2　用 `enumerate()` 输出序号

`enumerate(可迭代对象, start=1)` 在遍历时额外产出序号，`start` 控制起始值。

````python
print("序号从 1 开始:")
for idx, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"  {idx}. {name:<6} {score}")

print("\n（对照）start=0 时:")
for idx, name in enumerate(names[:3], start=0):
    print(f"  {idx}. {name}")
````

## 3.3　列表推导式

列表推导式 `[表达式 for 元素 in 可迭代对象 if 条件]` 一行完成「筛选 + 变换」。

````python
# 及格学生的姓名（成绩 >= 60）
passed_names = [name for name, score in zip(names, scores) if score >= 60]
print("及格学生:", passed_names)

# 所有成绩的平方
squared_scores = [score ** 2 for score in scores]
print("成绩平方:", squared_scores)

# 对照：用普通循环实现同样的结果
passed_loop = []
for name, score in zip(names, scores):
    if score >= 60:
        passed_loop.append(name)
print("\n普通循环结果:", passed_loop, " 与推导式一致?", passed_loop == passed_names)
````

## 3.4　基础统计与字符串格式化

| 指标 | 计算方式 |
|---|---|
| 人数 | `len(scores)` |
| 平均分 | `sum(scores) / len(scores)` |
| 最高分 | `max(scores)` |

````python
count = len(scores)
average = sum(scores) / count
highest = max(scores)
lowest = min(scores)
passed_count = sum(1 for s in scores if s >= 60)
pass_rate = passed_count / count

print(f"人数    : {count}")
print(f"平均分  : {average:.2f}")
print(f"最高分  : {highest}")
print(f"最低分  : {lowest}")
print(f"及格人数: {passed_count}")
print(f"及格率  : {pass_rate:.1%}")

# 补充：及格率不同的格式化写法
print("\n及格率的几种写法:")
print(f"  .1%   -> {pass_rate:.1%}")
print(f"  .2%   -> {pass_rate:.2%}")
print(f"  .0%   -> {pass_rate:.0%}")
print(f"  手算   -> {pass_rate * 100:.1f}%")
````

## 3.5　`alias = scores` 与 `copied = scores.copy()` 的区别

- `alias = scores`：只是给同一个列表对象**再起一个名字**，`alias is scores` 为 `True`
- `scores.copy()`：创建**新的列表对象**（浅拷贝），两者互不影响

本节使用**单独的演示列表**，避免污染后面实验要用的原始 `scores`。

````python
# 使用单独的演示列表，不碰原始 scores
demo = [85, 58, 92]
alias = demo            # 直接赋值：同一个对象的两个名字
copied = demo.copy()    # 浅拷贝：新对象

print("修改前:")
print("  demo  :", demo)
print("  alias :", alias)
print("  copied:", copied)

alias[0] = 0            # 通过 alias 修改

print("\n修改 alias[0] = 0 之后:")
print("  demo  :", demo,     "  <- 跟着变了")
print("  alias :", alias,    "  <- 跟着变了")
print("  copied:", copied,   "  <- 没变")

print("\n身份比较:")
print("  alias is demo :", alias is demo)
print("  copied is demo:", copied is demo)

print("\n原始 scores 是否受影响:", scores)
````

## 3.6　元组的不可变性 & 长度不同时的 `zip()`

- `tuple` 不可变：`record[0] = ...` 抛 `TypeError`，要「修改」只能构造**新元组**
- `zip()` 以**最短**序列为准，多出来的元素被忽略；需要保留时用 `itertools.zip_longest`

````python
# 用元组保存一条学生记录
record = ("Alice", 85)
print("记录:", record, " 类型:", type(record).__name__)

# 验证元素不能直接重新赋值
try:
    record[0] = "Bob"
except TypeError as e:
    print("捕获到 TypeError:", e)
    print("原因：tuple 是不可变类型，不支持元素赋值")

# 构造新元组
new_record = ("Bob",) + record[1:]
print("\n通过拼接构造新元组:", new_record)
print("原元组未变:", record)

# 元组也能解包
rec_name, rec_score = record
print("解包:", rec_name, rec_score)
````

````python
# 长度不同的两个序列使用 zip()
a = [1, 2, 3, 4]
b = ["a", "b"]

zipped = list(zip(a, b))
print("a =", a)
print("b =", b)
print("list(zip(a, b)) =", zipped)
print("长度:", len(zipped), " -> 以最短序列为准，a 中多出的 3、4 被丢弃")

# 想保留全部元素时用 zip_longest
from itertools import zip_longest
longest = list(zip_longest(a, b, fillvalue=None))
print("\nzip_longest(a, b, fillvalue=None) =", longest)
````

````python
import os
import sys


def locate_notebook_dir(notebook_name):
    """定位 notebook 所在目录，返回绝对路径；找不到则返回 None。"""
    here = os.path.abspath(os.getcwd())

    # 情况 1：当前目录就是 notebook 所在目录
    if os.path.exists(os.path.join(here, notebook_name)):
        return here

    # 情况 2：当前目录是 notebook 所在目录的父目录
    try:
        for entry in os.listdir(here):
            candidate = os.path.join(here, entry)
            if os.path.isdir(candidate) and os.path.exists(os.path.join(candidate, notebook_name)):
                return candidate
    except OSError:
        pass

    # 情况 3：向上逐级查找
    current = here
    while True:
        if os.path.exists(os.path.join(current, notebook_name)):
            return current
        parent = os.path.dirname(current)
        if parent == current:
            return None
        current = parent


NOTES = "20244080125_曹俊华_exp01.ipynb"
target_dir = locate_notebook_dir(NOTES)

if target_dir is None:
    print("警告：未能定位 notebook 所在目录，将使用当前目录：", os.getcwd())
elif target_dir != os.path.abspath(os.getcwd()):
    os.chdir(target_dir)
    print("已切换到 notebook 所在目录:", target_dir)
else:
    print("工作目录正确:", target_dir)
````

## 4.3　`parse_student(line)` 函数

依次做四道校验，任何一道不过就抛 `ValueError`，并把原因写清楚：

1. **字段数量**必须是 2（`name,score`）
2. **姓名不能为空**
3. **成绩必须能转成整数**
4. **成绩必须在 0~100 之间**

把「转整数」也纳入校验，是因为 `int("abc")` 本身会抛 `ValueError`，
统一成同一种异常类型，调用方只需捕获 `ValueError` 即可。

````python
def parse_student(line):
    """把一行 "name,score" 解析成 {"name": ..., "score": ...}。

    无效记录一律抛出 ValueError，并说明无效原因。
    """
    parts = line.strip().split(",")

    # 校验 1：字段数量
    if len(parts) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(parts)}")

    name = parts[0].strip()
    raw_score = parts[1].strip()

    # 校验 2：姓名非空
    if not name:
        raise ValueError("姓名不能为空")

    # 校验 3：成绩必须能转成整数
    try:
        score = int(raw_score)
    except ValueError:
        raise ValueError(f"成绩不是合法整数: {raw_score!r}")

    # 校验 4：成绩范围
    if not (0 <= score <= 100):
        raise ValueError(f"成绩超出 0~100 范围: {score}")

    return {"name": name, "score": score}
````

````python
# 单独测试这个函数
test_lines = ["Alice,85", "Bob,58", "Grace,abc", "Henry,105", "Incomplete", ",90", "  Eva , 60 "]
for line in test_lines:
    try:
        print(f"{line!r:<16} -> {parse_student(line)}")
    except ValueError as e:
        print(f"{line!r:<16} -> ValueError: {e}")
````

## 4.4　按行处理文件，捕获异常并继续

关键点：`try/except` 放在**循环内部**，这样单行出错不会中断整个文件；
把行号和原始内容一起记录下来，便于定位问题。

````python
valid_records = []   # 有效记录：字典组成的列表
errors = []          # 无效记录：(行号, 原始内容, 错误原因)

with open(DATA_FILE, "r", encoding="utf-8") as f:
    next(f)                                       # 跳过表头
    for line_no, line in enumerate(f, start=2):   # 数据从第 2 行开始
        if not line.strip():                      # 跳过空行
            continue
        try:
            valid_records.append(parse_student(line))
        except ValueError as e:
            errors.append((line_no, line.rstrip("\n"), str(e)))

print(f"有效记录数: {len(valid_records)}")
for r in valid_records:
    print("  ", r)

print(f"\n无效记录数: {len(errors)}")
for line_no, raw, reason in errors:
    print(f"  第{line_no}行 {raw!r:<16} -> {reason}")
````

| 模式 | 行为 |
|---|---|
| `"w"` | **覆盖**写入：文件原有内容被清空，从头写 |
| `"a"` | **追加**写入：保留原有内容，新内容写到末尾 |

下面分别演示两种模式的实际效果，最后写回完整报告。

````python
# 计算统计量
valid_scores = [r["score"] for r in valid_records]
valid_count = len(valid_records)
invalid_count = len(errors)
avg_score = sum(valid_scores) / valid_count
valid_pass_count = sum(1 for s in valid_scores if s >= 60)
valid_pass_rate = valid_pass_count / valid_count

REPORT_FILE = os.path.join(DATA_DIR, "report.txt")


def read_text(path):
    """读取文本文件内容（小工具函数）。"""
    with open(path, "r", encoding="utf-8") as fh:
        return fh.read()


def build_report():
    """构造完整的报告文本。"""
    return (
        "=== 实验 01 统计报告 ===\n"
        f"有效记录数: {valid_count}\n"
        f"无效记录数: {invalid_count}\n"
        f"平均分    : {avg_score:.2f}\n"
        f"及格率    : {valid_pass_rate:.1%}\n"
    )


print(build_report())
````

## 5.1　定义 `Student` 类

- `__init__()` 是构造方法，`self` 指向实例本身
- `is_passed()` 是普通方法，返回布尔值
- `__str__()` 决定 `print(对象)` 时显示什么
- 顺手实现 `__repr__()`，方便在列表里查看

````python
class Student:
    """学生对象：姓名 + 成绩。"""

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        """成绩 >= 60 视为及格。"""
        return self.score >= 60

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return f"Student(name={self.name!r}, score={self.score}, {status})"

    def __repr__(self):
        return f"Student({self.name!r}, {self.score})"
````

## 5.3　`GraduateStudent` 继承 `Student`

- `class GraduateStudent(Student):` 表示继承
- `super().__init__(name, score)` 复用父类的初始化逻辑
- 重写 `__str__()`，同时仍然可以复用父类的 `is_passed()`

````python
class GraduateStudent(Student):
    """研究生：在学生基础上增加研究方向。"""

    def __init__(self, name, score, research_topic):
        super().__init__(name, score)      # 初始化父类属性
        self.research_topic = research_topic

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return (f"GraduateStudent(name={self.name!r}, score={self.score}, "
                f"research_topic={self.research_topic!r}, {status})")


gs = GraduateStudent("Alice", 85, "机器学习")
print(gs)
print("\nis_passed() 继承自父类:", gs.is_passed())
print("是 GraduateStudent 实例?", isinstance(gs, GraduateStudent))
print("是 Student 实例?       ", isinstance(gs, Student))
print("属性字典:", gs.__dict__)
print("方法解析顺序 (MRO):", [c.__name__ for c in GraduateStudent.__mro__])
````

## 5.4　直接赋值 / `copy.copy()` / `copy.deepcopy()` 的差异

对象是 `{"name": "Alice", "scores": [85, 90]}`，内部**嵌套了一个列表**。

| 方式 | 新对象？ | 内部列表共享？ | 改内部列表的影响 |
|---|---|---|---|
| `alias = obj` | 否（同一个对象） | 是 | 原对象跟着变 |
| `copy.copy(obj)` | 是（浅拷贝） | **是** | 原对象跟着变 |
| `copy.deepcopy(obj)` | 是（深拷贝） | 否 | 原对象不受影响 |

每次比较都从**新建的原始对象**开始，避免上一次的修改串到下一次。

````python
import copy


def make_original():
    """每次调用都返回一个全新的原始对象（含嵌套列表）。"""
    return {"name": "Alice", "scores": [85, 90]}


print("原始对象工厂产物:", make_original())
````

````python
# ---- 1) 直接赋值 ----
original = make_original()
alias = original                     # 只是同一个对象的另一个名字

alias["scores"].append(100)          # 修改内部列表

print("【直接赋值 alias = original】")
print("  original:", original)
print("  alias   :", alias)
print("  alias is original          ->", alias is original)
print("  alias['scores'] is original['scores'] ->", alias["scores"] is original["scores"])
print("  >>> 同一个对象，改一处两处都变")
````

````python
# ---- 2) 浅拷贝 copy.copy() ----
original = make_original()           # 重新新建，避免受上一次影响
shallow = copy.copy(original)

shallow["scores"].append(100)        # 修改内部列表

print("【浅拷贝 shallow = copy.copy(original)】")
print("  original:", original)
print("  shallow :", shallow)
print("  shallow is original          ->", shallow is original)
print("  shallow['scores'] is original['scores'] ->", shallow["scores"] is original["scores"])
print("  >>> 外层是新字典，但内部列表仍然共享，所以 original 也被改了")
````

````python
# ---- 3) 深拷贝 copy.deepcopy() ----
original = make_original()           # 再次重新新建
deep = copy.deepcopy(original)

deep["scores"].append(100)           # 修改内部列表

print("【深拷贝 deep = copy.deepcopy(original)】")
print("  original:", original)
print("  deep    :", deep)
print("  deep is original          ->", deep is original)
print("  deep['scores'] is original['scores'] ->", deep["scores"] is original["scores"])
print("  >>> 连内部列表也复制了一份，改 deep 完全不影响 original")
````

````python
# 汇总对比表
print(f"{'方式':<14}{'外层同一对象':<14}{'内层同一对象':<14}{'改内层是否影响原对象'}")
print("-" * 62)

for label, maker in [
    ("直接赋值", lambda o: o),
    ("copy.copy", copy.copy),
    ("copy.deepcopy", copy.deepcopy),
]:
    original = make_original()
    other = maker(original)
    outer_same = other is original
    inner_same = other["scores"] is original["scores"]
    other["scores"].append(100)
    affected = original["scores"] != [85, 90]
    print(f"{label:<14}{str(outer_same):<14}{str(inner_same):<14}{affected}")
````
